```python
import pandas as pd
import matplotlib.pyplot as plt
flights = pd.read_csv('atlanta-airport-flights-2023.csv')
flights = flights.convert_dtypes()
flights
```

The `scheduled` column represents the time the flight was scheduled, but the actual departure time is represented by the `actual` column. null We can use these columns to calculate flight delays.

```python
# .copy() ensures departures is an independent DataFrame, not a view into
# `flights` -- without it, pandas raises a SettingWithCopyWarning as soon as
# we start assigning new columns to departures in the next few cells.
departures = flights[['scheduled', 'actual']].copy()
departures #display the dataframe
```

What data types are being used for these departure times? Let's find out using the `info()` method. Run the code below.

We can see there has been a slight formatting change.

Let's use `info()` to confirm the new data type. Run the code below.

```python
departures.info()
```

# Calculate the delays

To identify the delayed flights, let's first calculate the amount of delay for each flight using our datetime columns. The equation for the amount of flight delay is:

`delay = actual − scheduled`

Activity Goals:
- Calculate the amount of delay using eval().
- Assign the result to a new column.
- Display the dataframe.

```python
departures['delay'] = departures.eval('actual - scheduled')
departures #display
```

We have a new `delay` column! The first flight in our dataframe left 26 minutes late. The second flight has a negative delay, indicating that it left 5 minutes early.

Airlines can leave a few minutes behind schedule without being considered late. Typically a flight is considered late if it departs more than 900 seconds (15 minutes) after the scheduled departure time.

Let's determine which flights are late using `dt.total_seconds()`.

Activity Goals:
- Create a new column to flag flights as late.
- Display the dataframe.

**Why 900 seconds?** The 15-minute mark isn't arbitrary -- it's the U.S. Department of Transportation's official definition of a "late" flight, used in their own on-time performance reporting. Using the same threshold makes our numbers comparable to published DOT statistics.

**A formatting quirk to note:** pandas displays negative timedeltas in an unintuitive way. A flight that left 5 minutes early doesn't show `-0 days 00:05:00` -- it shows `-1 days +23:55:00` (mathematically equivalent, since -5 minutes equals -1 day + 23 hours 55 minutes). It's not a bug, just how pandas formats negative `Timedelta` values. We don't need to fix it, since `is_late` in the next cell relies on the underlying numeric seconds (`.dt.total_seconds()`) rather than the printed string.

```python
departures['is_late'] = departures['delay'].dt.total_seconds() > 900
departures # display dataframe
```

We now know which flights were delayed. Next up, calculating the day of week for each flight.

### Get the day of the week
In addition to supporting calculations, datetime objects allow us to get information such as the year, month, and day.

To do this, we take each datetime object and convert it to a formatted string using the string format time method, `strftime()`. The method uses format codes to represent the date and time. Below are a few examples:

`%d` - Day of the month as a zero-padded decimal number <br>
`%m` - Month as a zero-padded decimal number <br>
`%Y` - Year with century as a decimal number <br>

Activity Goals:
- Get the abbreviated weekday name for each flight.
- Assign the result to a new column.
- Display the dataframe.

```python
departures['day_name'] = departures['actual'].dt.strftime('%a')
departures #display
departures.to_csv('departures-check-point.tsv', index=False, sep='\t')
```

Each abbreviated day of the week is composed of three letters.

Now that we have the days of the week, let's calculate the percentage of flights that are delayed for each day of the week.

### Percentage flights delayed by day of week

To calculate the percentage of flights that are delayed, let's take the `mean()` of the `is_late` column for each day of the week. The mean of a True/False column equals the proportion of True values, as shown here.

We will then convert this proportion to a percentage by multiplying by 100.

Activity Goals:
- Group dataframe by day of the week.
- Calculate the mean of the 'is_late' column for each day.
- Convert that proportion to a percentage.
- Display the resulting series.

```python
proportion_delayed = departures.groupby('day_name')['is_late'].mean()
percent_delayed = proportion_delayed * 100
percent_delayed #Display series
```

The index for the series is day_name. null Notice the series is ordered by the index alphabetically. From this we can see that the day with the highest percentage of delays is Sunday.

These differences will be easier to see in a graph, so let's create a bar chart.

### Plot the data
We are almost there! The percent_delayed series is ordered alphabetically. To plot the data, let's first reorder percent_delayed in natural day of the week order (Sun-Sat).

Run the code below, which uses the `reindex( )` method.

```python
new_index_order = ['Sun', 'Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat']
percent_delayed = percent_delayed.reindex(new_index_order)
percent_delayed
```

Percent delayed is now ordered from Sunday through Saturday. We have a clean dataset that we can plot.

Activity Goals:
- Plot the bar chart.
- Label the y axis.

```python
plt.figure(figsize=(6, 3))
plt.bar(percent_delayed.index, percent_delayed)
plt.ylabel('Percent Delayed')
plt.show()
```

The day of the week with the fewest late flights is Tuesday.

```python
passengers = pd.read_csv('us-daily-passengers.csv')
passengers.head()
```

### Convert dates and get day of week

Just like we did for the flight data, we need to convert the `date` column to a proper datetime object before we can extract the day of the week.

```python
passengers['date'] = pd.to_datetime(passengers['date'])
passengers['day_name'] = passengers['date'].dt.strftime('%a')
passengers.head()
```

### Average passengers per day of week

Now we group by `day_name` and take the mean number of passengers, then reorder the results Sunday through Saturday so it lines up with our delay chart.

```python
avg_passengers = passengers.groupby('day_name')['num_passengers'].mean()
avg_passengers = avg_passengers.reindex(new_index_order)
avg_passengers
```

### Plot passenger volume by day of week

```python
plt.figure(figsize=(6, 3))
plt.bar(avg_passengers.index, avg_passengers, color='darkorange')
plt.ylabel('Average Passengers')
plt.title('Average Daily TSA Passenger Volume by Day of Week (2023)')
plt.show()
```

### Side-by-side comparison

Let's put both charts on the same figure so the patterns are easier to compare.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.5))

axes[0].bar(percent_delayed.index, percent_delayed, color='steelblue')
axes[0].set_ylabel('Percent Delayed')
axes[0].set_title('ATL Departure Delays by Day')

axes[1].bar(avg_passengers.index, avg_passengers, color='darkorange')
axes[1].set_ylabel('Average Passengers')
axes[1].set_title('US Passenger Volume by Day')

plt.tight_layout()
plt.show()
```

### What patterns do we see?

Tuesday is the low point for both metrics: it has the fewest average passengers nationwide (~2.10M) and the lowest delay rate at ATL (~15.4%). Friday and Sunday sit at the other end -- among the highest passenger counts (~2.55M and ~2.51M) and among the highest delay rates (~22.6% and ~23.4%). Saturday, interestingly, is low on both (~2.17M passengers, ~18.6% delayed), which doesn't fit a simple "weekend = busier" story, but does fit the pattern that days with fewer passengers tend to have fewer delays.

This is suggestive of a link between how many people are flying and how often flights run late (more passengers and more connecting traffic can mean more congestion at gates, security, and in the air), but it's **not proof of causation**. We're comparing a nationwide passenger metric to a single-airport delay metric, and both could be driven by a third factor, like weather patterns or holiday travel clustering, that happens to follow a similar weekly rhythm. A more rigorous next step would be to compare *ATL-specific* passenger counts to *ATL-specific* delays, and to control for seasonal effects (e.g. holidays) separately from day-of-week effects.

## Limitations

- **Sample size:** the flight delay analysis uses a random sample of 5,000 flights out of 336,434 total ATL departures in 2023 (about 1.5%). The day-of-week percentages are estimates, not exact population values, and carry some sampling error.
- **Single airport:** all delay data comes from Atlanta (ATL). Day-of-week delay patterns at other airports, especially smaller ones or those with different weather/traffic profiles, may look different.
- **Single year:** 2023 may have had atypical events (weather, holiday timing, staffing) that don't generalize to other years.
- **Correlation, not causation:** the passenger-volume comparison in the extension section shows two metrics that move together across the week, but doesn't establish that higher passenger volume *causes* more delays at ATL specifically, since the two datasets cover different scopes (nationwide vs. one airport).
